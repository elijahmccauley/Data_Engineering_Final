## Step 1: Setup

```python
import os
import re
import pandas as pd
from dotenv import load_dotenv
from openai import OpenAI
from autoddg import AutoDDG

# 1. Authenticate
load_dotenv(".secrets")
api_key = os.getenv("OPENAI_API_KEY")
if not api_key:
    print("Error: OPENAI_API_KEY not found in .secrets file.")
else:
    client = OpenAI(api_key=api_key)
    print("Authenticated successfully!")

# 2. Define Helper Functions
def word_count(text):
    return len(re.findall(r"\b\w+\b", text))

def force_exact_words(client, text, target_words=200, model="gpt-4o-mini", max_tries=5):
    current = text
    for _ in range(max_tries):
        response = client.chat.completions.create(
            model=model,
            temperature=0,
            messages=[
                {"role": "system", "content": f"Rewrite the dataset description to be exactly {target_words} words. Keep it factual, concise, and based only on the provided text. Do not add unsupported details."},
                {"role": "user", "content": current}
            ]
        )
        current = response.choices[0].message.content.strip()
        if word_count(current) == target_words:
            return current
    return current
```

## Step 2: Load Data

```python
fashion_path = "./amazon_clothes/fashion.csv"
amazon_tech_path = "./amazon_tech/amazon_tech.csv" 
gingivitis_images_path = "./dental_dataset/Dataset/Training/Train_captions.csv"

# Load dataframes
fashion_df = pd.read_csv(fashion_path)
amazon_df = pd.read_csv(amazon_tech_path)
gingivitis_images_df = pd.read_csv(gingivitis_images_path)

print("Baseline data loaded successfully!")
```

## Step 3 - Baseline with AutoDDG

### fashion_df

```python
sample_csv = fashion_df.sample(n=min(200, len(fashion_df)), random_state=27).to_csv(index=False)

autoddg = AutoDDG(
    client=client,
    model_name="gpt-4o-mini",
    description_words=200
)

_, description = autoddg.describe_dataset(dataset_sample=sample_csv)

if word_count(description) != 200:
    description = force_exact_words(client, description, target_words=200)

print(description)
```
